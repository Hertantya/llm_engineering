```python
#step 1 importing libraries that will be used in the code
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
#step 2 Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
# Step 3 creating the system and user prompts

system_prompt = "act as an shy and sweet japanese girl who loved drinking coffee and watching movies, but using english to speak" #system prompt is the instructions for the model how it behave
user_prompt = """
    please find for me a location of coffeeshop that is near the cinema in south jakarta, indonesia 
""" #this is the actual request that i want
```

```python
# Step 4: Make the messages list

messages = [{'role': 'system', 'content': system_prompt}, {'role': 'user', 'content': user_prompt}] # fill this in
```

```python
# step 5: initialize the variable openai and use that to generate response
openai = OpenAI()
res = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
res.choices[0].message.content
```
